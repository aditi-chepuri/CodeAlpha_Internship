```python
import pandas as pd

df1 = pd.read_csv("Unemployment in India.csv")
df2 = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")

print("Dataset 1:")
display(df1.head())

print("Dataset 2:")
display(df2.head())
```

```python
print("Dataset 1 Shape:", df1.shape)
print("Dataset 2 Shape:", df2.shape)

print("\nDataset 1 Columns:")
print(df1.columns.tolist())

print("\nDataset 2 Columns:")
print(df2.columns.tolist())
print("\nMissing values in Dataset 1:")
print(df1.isnull().sum())

print("\nMissing values in Dataset 2:")
print(df2.isnull().sum())
```

```python
# Remove completely empty rows
df1 = df1.dropna(how='all')

# Remove extra spaces from column names
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

print("Dataset 1 shape after cleaning:", df1.shape)
print("Dataset 2 shape:", df2.shape)

print("\nDataset 1 columns:")
print(df1.columns.tolist())
```

```python
df1.head()
```

```python
df1['Date'] = pd.to_datetime(df1['Date'], dayfirst=True)
df2['Date'] = pd.to_datetime(df2['Date'], dayfirst=True)

print(df1.dtypes)
```

```python
df1.head()
```

```python
# Calculate average unemployment rate for each date
monthly_unemployment = df1.groupby('Date')['Estimated Unemployment Rate (%)'].mean()

# Plot the trend
plt.figure(figsize=(12, 6))

plt.plot(monthly_unemployment.index, monthly_unemployment.values)

plt.title('Unemployment Rate Trend in India')
plt.xlabel('Date')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df1['Date'],
    df1['Estimated Unemployment Rate (%)']
)

plt.title('Unemployment Rate in India Over Time')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
yearly_stats = df1.groupby('Year').agg({
    'Estimated Unemployment Rate (%)': 'mean',
    'Estimated Employed': 'mean',
    'Estimated Labour Participation Rate (%)': 'mean'
})

print(yearly_stats)
display(yearly_stats.round(2))
```

```python
correlation = df1[
    [
        'Estimated Unemployment Rate (%)',
        'Estimated Employed',
        'Estimated Labour Participation Rate (%)'
    ]
].corr()

display(correlation.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Between Employment Indicators')

plt.show()
```

```python
summary = {
    'Average Unemployment Rate (%)': df1['Estimated Unemployment Rate (%)'].mean(),
    'Maximum Unemployment Rate (%)': df1['Estimated Unemployment Rate (%)'].max(),
    'Minimum Unemployment Rate (%)': df1['Estimated Unemployment Rate (%)'].min(),
    'Average Estimated Employed': df1['Estimated Employed'].mean(),
    'Average Labour Participation Rate (%)': df1['Estimated Labour Participation Rate (%)'].mean()
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=['Metric', 'Value']
)

display(summary_df.round(2))
```

```python
yearly_covid = (
    df1.groupby(df1['Date'].dt.year)['Estimated Unemployment Rate (%)']
    .mean()
)

print(yearly_covid)
```

```python
change = yearly_covid.loc[2020] - yearly_covid.loc[2019]

percentage_change = (
    change / yearly_covid.loc[2019]
) * 100

print("2019 Average Unemployment Rate:",
      round(yearly_covid.loc[2019], 2), "%")

print("2020 Average Unemployment Rate:",
      round(yearly_covid.loc[2020], 2), "%")

print("Change:",
      round(change, 2), "percentage points")

print("Percentage Change:",
      round(percentage_change, 2), "%")
```

```python
yearly_covid = pd.DataFrame({
    'Year': [2019, 2020],
    'Unemployment Rate': [
        yearly_covid.loc[2019],
        yearly_covid.loc[2020]
    ]
})

display(yearly_covid.round(2))
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    yearly_covid['Year'].astype(str),
    yearly_covid['Unemployment Rate']
)

plt.title('Average Unemployment Rate: 2019 vs 2020')
plt.xlabel('Year')
plt.ylabel('Unemployment Rate (%)')
plt.grid(axis='y')

plt.show()
```

## Key Findings

1. The unemployment data shows noticeable changes in unemployment rates between 2019 and 2020.

2. The average unemployment rate increased from 9.4% in 2019 to 15.1% in 2020.

3. This represents an increase of 5.7 percentage points, or approximately 60.67% compared with the 2019 average.

4. The analysis shows that unemployment rates varied across different regions of India.

5. Rural and urban areas showed differences in their average unemployment rates.

6. The monthly trend shows considerable variation in unemployment rates during the analyzed period.

7. The state-wise analysis shows that unemployment was not uniform across India; different regions experienced different unemployment levels.

8. The 2020 data provides a useful view of unemployment during the COVID-19 period and helps identify major changes in the unemployment trend.

9. Employment and labour participation indicators also varied over the analyzed period, providing additional context for understanding unemployment.
